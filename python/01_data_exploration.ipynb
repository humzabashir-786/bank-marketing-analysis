### Customer Distribution by Occupation

**Insight**

Blue-collar workers represent the largest customer segment, followed by management and technicians. These occupations account for a substantial proportion of the bank's customer base and should be investigated further to determine whether they also have high subscription rates.

### Which occupations have the highest subscription rate?

```python
job_subscription = pd.crosstab(
    df["job"],
    df["subscribed"],
    normalize="index"
) * 100

job_subscription
```

```python
job_subscription["Yes"].sort_values().plot(
    kind="barh",
    figsize=(10,6)
)

plt.title("Subscription Rate by Occupation")
plt.xlabel("Subscription Rate (%)")
plt.ylabel("Occupation")
plt.show()
```

### Which month had the highest subscription rate?

```python
month_subscription = pd.crosstab(
    df["month"],
    df["subscribed"],
    normalize="index"
) * 100

month_subscription
```

```python
month_subscription["Yes"].sort_values().plot(
    kind="barh",
    figsize=(10,6)
)

plt.title("Subscription Rate by Month")
plt.xlabel("Subscription Rate (%)")
plt.ylabel("Month")
plt.show()
```

### Which month had the highest subscription rate?

**Insight**

Subscription rates varied considerably throughout the year. March achieved the highest subscription rate at just over 50%, followed by December, September and October. In contrast, May recorded the lowest subscription rate at approximately 7%. This suggests that the timing of marketing campaigns may have a significant impact on customer response and should be considered when planning future campaigns.

### Does having a housing loan affect subscription rates?

```python
housing_subscription = pd.crosstab(
    df["housing"],
    df["subscribed"],
    normalize="index"
) * 100

housing_subscription
```

```python
housing_subscription["Yes"].plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Subscription Rate by Housing Loan")
plt.xlabel("Housing Loan")
plt.ylabel("Subscription Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

### Does having a housing loan affect subscription rates?

**Insight**

Customers without a housing loan subscribed at more than twice the rate of customers with a housing loan (approximately 17% compared with 8%). This may indicate that customers without existing housing debt are more willing or financially able to invest in term deposits.

### Does customer age influence subscription?

```python
df.groupby("subscribed")["age"].mean()
```

```python
df.groupby("subscribed")["age"].mean().plot(
    kind="bar",
    figsize=(6,4)
)

plt.title("Average Age by Subscription Status")
plt.xlabel("Subscription")
plt.ylabel("Average Age")
plt.xticks(rotation=0)
plt.show()
```

### Does customer age influence subscription?

**Insight**

Customers who subscribed were, on average, slightly older than those who did not. While the difference is relatively small, it suggests that older customers may be marginally more receptive to term deposit products.
